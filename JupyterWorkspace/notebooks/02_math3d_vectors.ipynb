# Math3D: Vectors and Rotation

Explore dot products, cross products, normalization, and rotation around the Z axis.

```python
import numpy as np
import matplotlib.pyplot as plt

a = np.array([2.0, 1.0, 0.5])
b = np.array([-0.5, 1.5, 2.0])
dot = np.dot(a, b)
cross = np.cross(a, b)
angle = np.degrees(np.arccos(dot / (np.linalg.norm(a) * np.linalg.norm(b))))
{"dot": dot, "cross": cross, "angle_degrees": angle}
```

```python
theta = np.radians(35)
rotation_z = np.array([
    [np.cos(theta), -np.sin(theta), 0],
    [np.sin(theta),  np.cos(theta), 0],
    [0, 0, 1],
])
rotated_a = rotation_z @ a
rotated_a
```

```python
fig = plt.figure(figsize=(7, 6))
ax = fig.add_subplot(111, projection="3d")
vectors = [(a, "a", "#235d77"), (b, "b", "#db6b2d"), (cross, "a x b", "#4f7d52"), (rotated_a, "Rz(a)", "#8b4f73")]
for vector, label, color in vectors:
    ax.quiver(0, 0, 0, *vector, color=color, linewidth=2, label=label)
limit = max(np.abs(np.concatenate([item[0] for item in vectors]))) + 0.5
ax.set(xlim=(-limit, limit), ylim=(-limit, limit), zlim=(-limit, limit), xlabel="X", ylabel="Y", zlabel="Z")
ax.legend()
plt.show()
```
