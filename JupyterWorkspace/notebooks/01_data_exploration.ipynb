# Data Exploration

Generate a deterministic experiment table, summarize it, and visualize the relationship between input size and runtime.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
sizes = np.repeat([100, 250, 500, 1000], 8)
runtime_ms = 0.018 * sizes + rng.normal(0, 1.8, sizes.size)
frame = pd.DataFrame({"size": sizes, "runtime_ms": runtime_ms})
frame.head()
```

```python
summary = frame.groupby("size")["runtime_ms"].agg(["mean", "std", "min", "max"])
summary.round(2)
```

```python
fig, ax = plt.subplots(figsize=(8, 4.5))
ax.scatter(frame["size"], frame["runtime_ms"], alpha=0.6, color="#235d77")
ax.plot(summary.index, summary["mean"], color="#db6b2d", linewidth=3, label="mean")
ax.set(title="Runtime scaling", xlabel="Input size", ylabel="Runtime (ms)")
ax.grid(alpha=0.2)
ax.legend()
plt.show()
```
