## 2. Build small reusable functions

Keep one idea per cell. This makes failures easy to locate and explanations easy to follow.

```python
def summarize(values):
    return {
        "count": len(values),
        "minimum": min(values),
        "maximum": max(values),
        "mean": sum(values) / len(values),
    }

summarize(numbers)
```
