# Symbolic Mathematics

Use SymPy when exact expressions matter more than floating-point approximations.

```python
import sympy as sp
sp.init_printing()
x, y = sp.symbols("x y", real=True)
expression = (x**2 - 1) / (x - 1)
sp.simplify(expression)
```

```python
f = sp.exp(-x**2) * sp.sin(3*x)
derivative = sp.diff(f, x)
integral = sp.integrate(x**2 * sp.exp(-x), (x, 0, sp.oo))
limit = sp.limit(sp.sin(x) / x, x, 0)
derivative, integral, limit
```

```python
solutions = sp.solve([
    sp.Eq(2*x + y, 7),
    sp.Eq(x - y, 2),
], [x, y], dict=True)
solutions
```

Try changing the equations, then use sp.plot(...) or convert symbolic expressions into numerical functions with sp.lambdify(...).
